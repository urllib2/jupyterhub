# Lesson 1 — Your First ROS2 Commands
**Objectives:** understand nodes, topics, and publish/subscribe — using only the command line
**Time:** 30 min

📤 Kernel → Restart & Run All → Submit


---
## 1.0 — Before the code: what even is a "node"?

Imagine three people in a room. One keeps announcing the time, out loud,
every second — whether or not anyone's listening. Another person, if they
care, just... listens. They never talked to each other directly.
Nobody agreed on anything in advance except: "time announcements happen
on this channel."

That's the entire idea behind ROS2. Programs (**nodes**) announce things
(**publish** to **topics**), and other programs listen (**subscribe**) —
with zero direct connection between them.

Let's see it happen, with commands ROS2 already gives you for free.


---
## 1.1 — Start a node

Open a terminal in JupyterHub (**File → New → Terminal**) and run:

```bash
ros2 run demo_nodes_cpp talker
```

Leave it running. You'll see it printing "Publishing: 'Hello World: N'"
every second. That's a node, doing its one job.


---
## 1.2 — Prove the node exists

Open a **second terminal** and run:

```bash
ros2 node list
```

You should see `/talker` — the system knows this program exists, from
anywhere, without you telling it directly.

Try:
```bash
ros2 node info /talker
```

This shows everything the node publishes, subscribes to, and offers as
services — a full profile of one running program.


---
## 1.3 — Find what it's shouting

```bash
ros2 topic list
```

You'll see `/chatter` among others. Now listen in:

```bash
ros2 topic echo /chatter
```

Messages should start streaming — the exact thing `/talker` is
broadcasting, live, in your terminal. Press `Ctrl+C` to stop watching
(the talker keeps running, unaffected).


---
## 1.4 — Add a second program that actually listens

Open a **third terminal**:

```bash
ros2 run demo_nodes_cpp listener
```

Watch it print "I heard: [Hello World: N]" — reacting to the same
`/chatter` topic, in real time. Two independent programs. No shared
code. Just a topic between them.


---
## 1.5 — Inspect the connection

```bash
ros2 topic hz /chatter
```
Shows you the actual publish rate (should be ~1 Hz).

```bash
ros2 topic info /chatter --verbose
```
Shows exactly who's publishing and who's subscribing, plus QoS
settings — the same information a graphical tool like `rqt_graph`
would show you, but in text. This matters more than it sounds: most
real robots run headless (no screen attached), so this is how
professionals actually debug systems in the field.
