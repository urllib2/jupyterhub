## 2.0 — Not everything should be a shout

Every topic you've used so far works the same way: a node publishes,
and anyone listening gets it — no confirmation, no answer, no waiting.
That's perfect for a stream of sensor readings or velocity commands.
It's a terrible fit for a question like *"is the battery charged?"* or
*"can you compute this for me right now?"*

That's what **services** are for. A service is a **request → response**
pattern:
- One node offers a service — it waits, ready to answer.
- Another node calls it, sends a request, and **waits** for the reply.
- The call doesn't return until an answer comes back.

This is the opposite of a topic's fire-and-forget nature. Let's see it
before writing any code.

Open a terminal and run:
```bash
ros2 service list
```
You'll see a handful of default services every node exposes (things
like parameter management) — a preview of what's coming.

- Confirm the service name matches exactly: `/add_two_ints`.
- The terminal command needs the full type path:
  `example_interfaces/srv/AddTwoInts` — a typo there gives a "service
  not found" style error, not a wrong answer.

## 2.2 — Compare it to a topic, on purpose

While your service server (2.1) is **not** running (let it finish or
stop it), open a terminal and try:

```bash
ros2 service call /add_two_ints example_interfaces/srv/AddTwoInts "{a: 3, b: 4}"
```

Notice what happens: the terminal just **hangs**, waiting. Press
`Ctrl+C` to stop it after a few seconds.

Now compare: back in Module 01, if you ran `ros2 topic pub` to
`/chatter` with no `listener` running, the command still completed
instantly — publishing doesn't care if anyone's there. Calling a
service with no server running just **waits**, because it's expecting
an answer that will never come.
