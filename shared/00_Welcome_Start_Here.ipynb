## 🤖 What is ROS 2?

**ROS 2** (Robot Operating System 2) is free, open-source software for building robots. Despite the name, it is not an operating system. It is a toolkit that runs on Linux.

A real robot is made of many small programs: one reads the sensors, one plans a path, one drives the motors. ROS 2 lets these programs run side by side and exchange information.



It comes with:
- **Command-line tools**, all starting with `ros2`
- **Libraries** to write your own robot programs in Python (`rclpy`) or C++

Each ROS 2 release has a name. This course uses **Jazzy**.

You see all of this in action in a few minutes. Nothing to memorize yet.

## 🐍 Check ROS 2 from Python

`rclpy` is the ROS 2 library for Python. This cell imports it, creates the smallest possible ROS 2 program, prints its name, and cleans up:

````python
import rclpy

rclpy.init()
node = rclpy.create_node("hello_rosforge")
print("Node created:", node.get_name())

node.destroy_node()
rclpy.shutdown()
````

**Expected output:**

```
Node created: hello_rosforge
```

That object is a **node**, the basic building block of ROS 2. You build many of them in lesson 01.

> A harmless `[WARN] ... ROS_LOCALHOST_ONLY is deprecated` message may appear. You can ignore it.

## 🚀 Do it — your first Terminal

A robot program keeps running until you stop it, so you start it in a **Terminal**. ROS 2 includes demo programs you can try right away.

### Step 1 — Start a program

Open a Terminal: **File → New → Terminal** (or the **Terminal** tile in the Launcher). Click inside it and run:

```bash
ros2 run demo_nodes_py talker
```

`ros2 run <package> <program>` starts a program. Here the package is `demo_nodes_py` and the program is `talker`.

A new line appears about every second:

```
[INFO] [...] [talker]: Publishing: 'Hello World: ...'
```

The program does not stop on its own, and the prompt does not come back. Leave it running.

### Step 2 — Look at it from a second Terminal

Open a **second Terminal** (File → New → Terminal) and drag its tab to the right side to see both at once. Run:

```bash
ros2 node list
```

You see:

```
/talker
```



`ros2 node list` shows the ROS 2 programs that are running right now. (Lesson 01 explains the word *node*.)

### Step 3 — Stop it

Go back to the **first** Terminal and press **Ctrl+C**. The program stops and the prompt returns. In the second Terminal, run `ros2 node list` again. The list is empty.

You will use this routine all course: **one Terminal per running program, and Ctrl+C to stop it.**
