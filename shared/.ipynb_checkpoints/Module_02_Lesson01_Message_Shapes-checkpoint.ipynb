## 1.0 — Not everything is a string

Every message you sent in Module 01 was `std_msgs/msg/String` — just
text. That's the simplest message type ROS2 has, and real robots barely
use it. What actually drives a robot is **structured** data: numbers
with names, grouped together.

Open a terminal and run:

```bash
ros2 interface show geometry_msgs/msg/Twist
```

You'll see something like:
```
Vector3  linear
	float64 x
	float64 y
	float64 z
Vector3  angular
	float64 x
	float64 y
	float64 z
```

That's a `Twist` — the message type that tells a robot how fast to
move. `linear.x` is forward/backward speed. `angular.z` is how fast to
turn. This exact message type drives the `/cmd_vel` topic on nearly
every mobile robot you'll ever work with — simulated or real.

Compare it to the one you already know:
```bash
ros2 interface show std_msgs/msg/String
```
One field, `data`, just text. `Twist` has two fields, each containing
three more fields inside it. That nesting is what "structured" means.

## 1.1 — Build a Twist publisher

Same pattern as Module 01: most of the code is done. Fill in the blanks.

**What you're filling in:**
1. `self.publisher_` — create a publisher for topic `/cmd_vel`, type
   `Twist`, queue size 10.
2. Inside the callback — set `msg.linear.x` to `0.5` (half a meter per
   second forward) and `msg.angular.z` to `0.2` (a slow turn).


````python
import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist
import time

class VelocityPublisher(Node):
    def __init__(self):
        super().__init__('velocity_publisher')
        # YOUR CODE HERE — create a publisher: topic '/cmd_vel', type Twist, queue size 10
        self.publisher_ = None
        self.timer = self.create_timer(1.0, self.timer_callback)

    def timer_callback(self):
        msg = Twist()
        # YOUR CODE HERE — set msg.linear.x = 0.5 and msg.angular.z = 0.2
        self.publisher_.publish(msg)
        self.get_logger().info(f'linear.x={msg.linear.x}  angular.z={msg.angular.z}')

### BEGIN SOLUTION

### END SOLUTION

# --- Clean up previous run (safe to re-run this cell any number of times) ---
if 'vel_node' in globals() and vel_node is not None:
    try:
        vel_node.destroy_node()
    except Exception:
        pass
if not rclpy.ok():
    rclpy.init()

vel_node = VelocityPublisher()

print("✅ Publisher running for 12 seconds.")
print("   Open a terminal NOW and run: ros2 topic echo /cmd_vel")
print("   You have 12 seconds — switch windows immediately.")

end_time = time.time() + 12
while time.time() < end_time and rclpy.ok():
    rclpy.spin_once(vel_node, timeout_sec=0.1)

vel_node.destroy_node()
print("✅ Done. Did you see linear/angular fields streaming in the terminal?")
````
