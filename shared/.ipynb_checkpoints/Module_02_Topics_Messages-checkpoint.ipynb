## 2.1 — The water pipe analogy

Imagine the /scan topic as a water pipe.
The robot pumps water (messages) at 10 litres per second.
Your processing node can only handle 5 litres per second.

**depth=1** — overflow is discarded immediately.
You always get the freshest water. Old water is thrown away.
This is correct for sensors.

**depth=100** — water queues up.
You process old water long after it arrived.
Your robot reacts to a sensor reading from 10 seconds ago.
This is dangerous for real-time control.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
fig.patch.set_facecolor('#1a1a2e')
for ax in [ax1, ax2]:
    ax.set_facecolor('#16213e'); ax.axis('off')
    ax.set_xlim(0,10); ax.set_ylim(0,10)

def pipe_scene(ax, depth, title, color, overflow):
    ax.add_patch(mpatches.FancyBboxPatch((0.3,6.5),2,2,
        boxstyle='round,pad=0.1',fc='#1565C0',ec='white',lw=2))
    ax.text(1.3,7.5,'ROBOT',ha='center',va='center',color='white',fontsize=10,fontweight='bold')
    ax.add_patch(mpatches.Rectangle((2.3,7.2),2,0.6,fc='#78909C'))
    ax.annotate('',xy=(4.3,7.5),xytext=(2.3,7.5),
                arrowprops=dict(arrowstyle='->',color='#42a5f5',lw=2))
    ax.text(3.3,8.0,'10 msg/s',ha='center',color='#42a5f5',fontsize=8)
    fill_h = min(depth*0.04, 3.5)
    ax.add_patch(mpatches.Rectangle((4.3,5.5),2,4,fc='none',ec=color,lw=3))
    ax.add_patch(mpatches.Rectangle((4.3,5.5),2,fill_h,fc=color,alpha=0.4))
    ax.text(5.3,7.5,f'depth={depth}',ha='center',va='center',color='white',fontsize=10,fontweight='bold')
    if overflow:
        ax.text(5.3,9.9,'overflow!',ha='center',color='#ef5350',fontsize=9,fontweight='bold')
    ax.add_patch(mpatches.FancyBboxPatch((7,6.5),2,2,
        boxstyle='round,pad=0.1',fc='#2e7d32',ec='white',lw=2))
    ax.text(8,7.5,'PROCESS\n5 msg/s',ha='center',va='center',color='white',fontsize=9,fontweight='bold')
    ax.annotate('',xy=(7,7.5),xytext=(6.3,7.5),
                arrowprops=dict(arrowstyle='->',color='#66bb6a',lw=2))
    ax.set_title(title,color='white',fontsize=12,pad=8)
    msg = '✅ Always fresh data' if not overflow else '⚠️  Processing stale data!'
    ax.text(5,0.5,msg,ha='center',
            color='#66bb6a' if not overflow else '#ef5350',fontsize=11,fontweight='bold')

pipe_scene(ax1,1,'depth=1 (sensor — correct)','#42a5f5',False)
pipe_scene(ax2,100,'depth=100 (sensor — WRONG)','#ef5350',True)
plt.suptitle('QoS Depth: how many messages to buffer',color='white',fontsize=14,fontweight='bold')
plt.tight_layout()
plt.show()
```

**Rule:** sensor topics → `depth=1`, `BEST_EFFORT`.
Command topics → `depth=10`, `RELIABLE`.

QoS mismatch: ROS2 drops messages **silently**. No error. No warning.

```python
import rclpy
from rclpy.node import Node
from rclpy.qos import QoSProfile, ReliabilityPolicy, HistoryPolicy
from sensor_msgs.msg import LaserScan
import numpy as np

sensor_qos = QoSProfile(
    reliability=ReliabilityPolicy.BEST_EFFORT,
    history=HistoryPolicy.KEEP_LAST, depth=1)

rclpy.init()
pub_node = Node('lidar_sim'); sub_node = Node('scan_proc')
scans_received = []

def process(msg):
    r = np.array(msg.ranges)
    valid = r[(r>0.12)&(r<3.5)]
    scans_received.append({'n': len(r), 'min': round(float(valid.min()),2) if len(valid) else 0})

pub = pub_node.create_publisher(LaserScan, '/scan_demo', sensor_qos)
sub = sub_node.create_subscription(LaserScan, '/scan_demo', process, sensor_qos)

for i in range(5):
    scan = LaserScan()
    scan.header.stamp = pub_node.get_clock().now().to_msg()
    scan.angle_min=-np.pi; scan.angle_max=np.pi
    scan.angle_increment=2*np.pi/360
    scan.range_min=0.12; scan.range_max=3.5
    scan.ranges=list(np.random.uniform(0.5,3.0,360))
    pub.publish(scan)
    rclpy.spin_once(sub_node, timeout_sec=0.05)

print(f"Published: 5 | Received: {len(scans_received)}")
for i,s in enumerate(scans_received):
    print(f"  Scan {i+1}: {s['n']} readings, min={s['min']}m")

pub_node.destroy_node(); sub_node.destroy_node(); rclpy.shutdown()
```

## 2.2 — Math: LaserScan geometry

Each reading $r_i$ corresponds to angle $\alpha_i = \alpha_{min} + i \cdot \Delta\alpha$

The Cartesian position of each obstacle:

$$x_i = r_i \cos(\alpha_i) \qquad y_i = r_i \sin(\alpha_i)$$

This is what the occupancy grid builder (Module 06) uses to update each cell.
