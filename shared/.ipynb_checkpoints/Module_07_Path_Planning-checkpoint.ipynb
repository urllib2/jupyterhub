```python
%matplotlib widget
import numpy as np, matplotlib.pyplot as plt
import heapq
from collections import defaultdict
import ipywidgets as widgets
from IPython.display import display

def make_map(sz=28):
    g=np.zeros((sz,sz))
    g[7,3:22]=1; g[17,6:25]=1; g[3:22,22]=1; g[12,3:16]=1
    return g

def nbrs8(grid,r,c):
    rows,cols=grid.shape; res=[]
    for dr in[-1,0,1]:
        for dc in[-1,0,1]:
            if dr==0 and dc==0: continue
            nr,nc=r+dr,c+dc
            if 0<=nr<rows and 0<=nc<cols and grid[nr,nc]==0:
                res.append((nr,nc,1.414 if dr!=0 and dc!=0 else 1.0))
    return res

def oct(a,b):
    dr,dc=abs(a[0]-b[0]),abs(a[1]-b[1])
    return max(dr,dc)+(1.414-1)*min(dr,dc)

def astar_h(grid,start,goal):
    pq=[(0,start[0],start[1])]; cf={start:None}
    g=defaultdict(lambda:float('inf')); g[start]=0; hist=[]
    while pq:
        f,r,c=heapq.heappop(pq); cur=(r,c); hist.append(cur)
        if cur==goal:
            path=[]; nd=goal
            while nd: path.append(nd); nd=cf[nd]
            return list(reversed(path)),hist
        for nr,nc,cost in nbrs8(grid,r,c):
            nb=(nr,nc); ng=g[cur]+cost
            if ng<g[nb]:
                g[nb]=ng; cf[nb]=cur
                heapq.heappush(pq,(ng+oct(nb,goal),nr,nc))
    return None,hist

GRID=make_map(); START=(1,1); GOAL=(26,26)
PATH,HIST=astar_h(GRID,START,GOAL)
fig,ax=plt.subplots(figsize=(9,9))
fig.patch.set_facecolor('#1a1a2e'); ax.set_facecolor('#16213e')
base=np.zeros((28,28,3))+0.08; base[GRID==1]=[0.3,0.3,0.3]
im=ax.imshow(base,origin='upper')
ax.plot(START[1],START[0],'g^',ms=14,label='Start',zorder=5)
ax.plot(GOAL[1],GOAL[0],'r*',ms=14,label='Goal',zorder=5)
ax.legend(facecolor='#1a1a2e',labelcolor='white')
sl=widgets.IntSlider(value=0,min=0,max=len(HIST)-1,
    description='Step:',layout=widgets.Layout(width='600px'))
def upd(chg):
    n=chg['new']; vis=base.copy()
    for r,c in HIST[:n]: vis[r,c]=[0.9,0.8,0.1]
    if PATH:
        for r,c in PATH: vis[r,c]=[0.1,0.4,1.0]
    im.set_data(vis)
    ax.set_title(f'A* step {n}/{len(HIST)} | path {len(PATH)} cells',color='white')
    fig.canvas.draw_idle()
sl.observe(upd,names='value'); display(sl); plt.show()
print(f'✅ A*: {len(PATH)} steps, {len(HIST)} cells explored')
print('Move slider to animate the search.')
```

## 7.1 — The hiking analogy

You are hiking in mountains. You want to reach the summit.

**Without a compass (Dijkstra):** you walk in every direction equally.
Left, right, back, forward — always in order of energy spent.
You find the optimal path, but you wander everywhere.

**With a compass (A*):** you always walk toward the summit.
Sometimes a cliff forces you to detour.
But you always trend toward the goal.
Same optimal path — far fewer locations explored.

The compass is the **heuristic** $h(n)$ — an estimate of remaining distance.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt, numpy as np, heapq
from collections import defaultdict

def dijkstra_h(grid,start,goal):
    pq=[(0,start[0],start[1])]; cf={start:None}
    g=defaultdict(lambda:float('inf')); g[start]=0; hist=[]
    while pq:
        cost,r,c=heapq.heappop(pq); cur=(r,c); hist.append(cur)
        if cur==goal:
            path=[]; nd=goal
            while nd: path.append(nd); nd=cf[nd]
            return list(reversed(path)),hist
        for nr,nc,mc in nbrs8(grid,r,c):
            nb=(nr,nc); ng=g[cur]+mc
            if ng<g[nb]:
                g[nb]=ng; cf[nb]=cur; heapq.heappush(pq,(ng,nr,nc))
    return None,hist

path_d,hist_d=dijkstra_h(GRID,START,GOAL)
path_a,hist_a=astar_h(GRID,START,GOAL)

fig,(ax1,ax2)=plt.subplots(1,2,figsize=(15,6))
fig.patch.set_facecolor('#1a1a2e')
for ax,path,hist,title in [
    (ax1,path_d,hist_d,f'Dijkstra\n{len(hist_d)} cells explored'),
    (ax2,path_a,hist_a,f'A* (octile heuristic)\n{len(hist_a)} cells explored')]:
    ax.set_facecolor('#16213e')
    vis=np.zeros((28,28,3))+0.08; vis[GRID==1]=[0.3,0.3,0.3]
    for r,c in hist[::2]: vis[r,c]=[0.9,0.8,0.1]
    if path:
        for r,c in path: vis[r,c]=[0.1,0.4,1.0]
    ax.imshow(vis,origin='upper')
    ax.plot(START[1],START[0],'g^',ms=12); ax.plot(GOAL[1],GOAL[0],'r*',ms=12)
    ax.set_title(title,color='white',fontsize=12); ax.axis('off')
imp=(len(hist_d)-len(hist_a))/len(hist_d)*100
fig.text(0.5,0.01,f'A* explores {imp:.0f}% fewer cells | Same optimal path',
         ha='center',color='#66bb6a',fontsize=12,fontweight='bold')
plt.tight_layout(); plt.show()
```

A* expands nodes by $f(n) = g(n) + h(n)$.
It stays focused on the goal. Same optimality guarantee as Dijkstra.

**Admissibility:** $h(n) \leq h^*(n)$ — never overestimate.
Octile distance is admissible for 8-connected grids.

## 7.2 — Math: octile heuristic

$$h_{octile}(a,b) = \max(\Delta r, \Delta c) + (\sqrt{2}-1)\min(\Delta r,\Delta c)$$

This is the **exact** cost on an open 8-connected grid.
With obstacles it underestimates — which preserves admissibility.

Manhattan distance $h = |\Delta r| + |\Delta c|$ is **not admissible** for 8-connected
because diagonal moves cost $\sqrt{2} < 2$.
