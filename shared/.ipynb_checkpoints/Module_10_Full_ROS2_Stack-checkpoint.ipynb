## Setup (run once on your local machine)

```bash
# Pull the dev container
docker pull urllib2/rosforge-tb3:latest

# VS Code: Ctrl+Shift+P → Dev Containers: Open Folder in Container

# Inside container terminal:
source /opt/ros/jazzy/setup.bash
export TURTLEBOT3_MODEL=burger
ros2 launch turtlebot3_gazebo turtlebot3_world.launch.py
```

---
## Verify topics are publishing

```bash
ros2 topic list
ros2 topic hz /scan      # should be ~10 Hz
ros2 topic hz /odom      # should be ~30 Hz
ros2 topic info /scan --verbose   # shows QoS profile
```

The QoS profile of /scan tells you which profile your subscriber must match.
This is the most common bug when connecting to TurtleBot3.
