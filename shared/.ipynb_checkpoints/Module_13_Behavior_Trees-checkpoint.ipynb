## Read the Nav2 BT

```bash
cat /opt/ros/jazzy/share/nav2_bt_navigator/behavior_trees/\
    navigate_w_replanning_and_recovery.xml
```

Key BT node types:
- `Sequence` — run children left→right, stop on first FAILURE
- `Fallback` — run children left→right, stop on first SUCCESS
- `RecoveryNode` — Sequence with fallback recovery on failure
- `ComputePathToPose` — calls your A* plugin (Module 12)
- `FollowPath` — DWB local planner + PID (Module 09)

---
## The decision logic

```xml
<RecoveryNode number_of_retries="6">
  <PipelineSequence>
    <RateController hz="1.0">
      <RecoveryNode>
        <ComputePathToPose/>    ← your A* (Module 12)
        <ClearEntireCostmap/>   ← recovery if planning fails
      </RecoveryNode>
    </RateController>
    <RecoveryNode>
      <FollowPath/>             ← DWB + PID (Module 09)
      <ClearEntireCostmap/>     ← recovery if following fails
    </RecoveryNode>
  </PipelineSequence>
  <Spin spin_dist="1.57"/>      ← last resort recovery
</RecoveryNode>
```

The robot:
1. Computes a path (A*) every second
2. Follows the path (DWB+PID) continuously
3. If either fails: clears costmap and retries
4. After 6 retries: spins 90° then tries again
