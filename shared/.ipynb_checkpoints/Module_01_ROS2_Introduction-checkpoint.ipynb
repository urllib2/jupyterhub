## 1.1 — The analogy before the code

Imagine you are building a robot with five components:
a lidar, a camera, a planner, a localizer, and two motors.

Without ROS2: every pair needs a custom protocol.
Five components → ten custom protocols. A maintenance nightmare.

ROS2 solves this with one idea: every component is a **node**.
Nodes communicate through **topics** using typed **messages**.

But there are two patterns. One is like a radio broadcast.
One is like a phone call.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.patch.set_facecolor('#1a1a2e')
for ax in [ax1, ax2]:
    ax.set_facecolor('#16213e')
    ax.set_xlim(0, 10); ax.set_ylim(0, 8)
    ax.axis('off')

# LEFT: Radio broadcast (topic)
ax1.add_patch(mpatches.FancyArrow(5,6,0,-2.5,width=0.15,
    head_width=0, fc='#42a5f5'))
ax1.add_patch(plt.Polygon([[4.7,6],[5.3,6],[5.15,7.2],[4.85,7.2]],
    fc='#42a5f5'))
for r, alpha in [(1.0,0.9),(1.8,0.6),(2.6,0.3)]:
    for t1,t2 in [(100,170),(10,80)]:
        arc = mpatches.Arc((5,6), 2*r, 2*r, angle=0,
                           theta1=t1, theta2=t2,
                           color='#42a5f5', alpha=alpha, lw=2)
        ax1.add_patch(arc)
for x,y in [(1.5,2),(4,1.5),(7,2),(8.5,4)]:
    ax1.add_patch(mpatches.Circle((x,y),0.35,fc='#66bb6a',zorder=3))
    ax1.annotate('',xy=(x,y),xytext=(5,3.5),
                 arrowprops=dict(arrowstyle='->',color='#42a5f5',alpha=0.5,lw=1.5))
ax1.text(5,0.3,'TOPIC — one publisher, many subscribers\nNo response. No confirmation.',
         ha='center',color='white',fontsize=10,fontweight='bold')
ax1.set_title('/scan, /cmd_vel, /odom',color='#42a5f5',fontsize=12,pad=8)

# RIGHT: Phone call (service)
ax2.add_patch(mpatches.FancyBboxPatch((1,3),3,2,
    boxstyle='round,pad=0.1',fc='#66bb6a',ec='white',lw=2))
ax2.add_patch(mpatches.FancyBboxPatch((6,3),3,2,
    boxstyle='round,pad=0.1',fc='#ef5350',ec='white',lw=2))
ax2.text(2.5,4,'CLIENT',ha='center',va='center',
         color='white',fontsize=13,fontweight='bold')
ax2.text(7.5,4,'SERVER',ha='center',va='center',
         color='white',fontsize=13,fontweight='bold')
ax2.annotate('',xy=(6,4.3),xytext=(4,4.3),
             arrowprops=dict(arrowstyle='->',color='#ffee58',lw=2.5))
ax2.text(5,4.65,'REQUEST',ha='center',color='#ffee58',fontsize=10,fontweight='bold')
ax2.annotate('',xy=(4,3.7),xytext=(6,3.7),
             arrowprops=dict(arrowstyle='->',color='#ff8a65',lw=2.5))
ax2.text(5,3.3,'RESPONSE',ha='center',color='#ff8a65',fontsize=10,fontweight='bold')
ax2.text(5,0.3,'SERVICE — request/response\nYou ask. You wait. You receive.',
         ha='center',color='white',fontsize=10,fontweight='bold')
ax2.set_title('/compute_path, /clear_costmap',color='#ef5350',fontsize=12,pad=8)

plt.suptitle('Two ways to communicate in ROS2',
             color='white',fontsize=14,fontweight='bold',y=1.01)
plt.tight_layout()
plt.show()
```

**Topic** = radio broadcast. One publisher, many subscribers. No response.
Best for continuous data: `/scan`, `/odom`, `/cmd_vel`.

**Service** = phone call. Request → wait → response.
Best for one-time queries: `/compute_path`, `/clear_costmap`.

**Rule:** need an answer → service. Need a data stream → topic.

## 1.2 — Your first node and publisher

```python
import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist

rclpy.init()
node = Node('rosforge_01')
pub = node.create_publisher(Twist, '/cmd_vel', 10)

msg = Twist()
msg.linear.x  = 0.3   # forward at 0.3 m/s
msg.angular.z = 0.0   # no rotation
pub.publish(msg)

print(f'✅ Node: {node.get_name()}')
print(f'   Published: linear.x={msg.linear.x}, angular.z={msg.angular.z}')
print()
print('Twist fields for differential drive:')
print('  linear.x  → forward/backward (m/s)')
print('  angular.z → rotation (rad/s)')
print('  All other fields → unused by TurtleBot3')

node.destroy_node()
rclpy.shutdown()
```

## 1.3 — Your first subscriber

```python
import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist

log = []

def callback(msg):
    log.append((round(msg.linear.x,3), round(msg.angular.z,3)))
    print(f'  📨 v={msg.linear.x:.2f} m/s  ω={msg.angular.z:.2f} rad/s')

rclpy.init()
pub_node = Node('pub'); sub_node = Node('sub')
pub = pub_node.create_publisher(Twist, '/motion', 10)
sub = sub_node.create_subscription(Twist, '/motion', callback, 10)

commands = [(0.3,0.0),(0.0,1.0),(0.3,0.5),(0.0,0.0)]
print('Sending commands:')
for vx, wz in commands:
    msg = Twist(); msg.linear.x=vx; msg.angular.z=wz
    pub.publish(msg)
    rclpy.spin_once(sub_node, timeout_sec=0.1)

print(f'\n✅ Sent {len(commands)}, received {len(log)}')
print('spin_once: processes ONE pending message and returns.')
print('Call it once per published message in notebooks.')

pub_node.destroy_node(); sub_node.destroy_node()
rclpy.shutdown()
```

## 1.4 — Math: the Twist message

For a differential drive robot, the velocity command is:

$$\mathbf{u} = \begin{bmatrix} v_x \\ \omega_z \end{bmatrix}$$

The robot's trajectory integrates this over time:

$$x(t) = \int_0^t v_x \cos\theta \, d\tau \quad
y(t) = \int_0^t v_x \sin\theta \, d\tau \quad
\theta(t) = \int_0^t \omega_z \, d\tau$$

This integration — velocity → position — is **dead reckoning** (Module 09).
