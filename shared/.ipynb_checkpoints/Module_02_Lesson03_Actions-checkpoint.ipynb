## 3.0 — When "wait for the answer" isn't enough

Services (Lesson 2) are great when the answer comes back instantly —
add two numbers, get a sum, done in milliseconds. But think about
telling a robot: *"navigate to the kitchen."* That could take thirty
seconds. A service call would just sit there, frozen, telling you
nothing until it's completely finished — no sense of progress, and no
way to change your mind halfway through.

That's what **actions** are for. An action is built from three parts:
- **Goal** — what you're asking for (e.g. "drive here")
- **Feedback** — repeated progress updates while it's working
- **Result** — the final outcome, once it's actually done

And critically: **you can cancel an action while it's running.** You
can't cancel a service call once it's sent.

Open a terminal and run:
```bash
ros2 run examples_rclpy_minimal_action_server server &
ros2 action list
```
Then in another terminal:
```bash
ros2 action send_goal /fibonacci example_interfaces/action/Fibonacci "{order: 5}" --feedback
```
Watch it print **feedback lines** as it works, before the final result
appears. That streaming feedback is the whole point — something a
service can never do.

## 3.1 — Build an action server

Below is an action server that counts down from a number, sending
feedback at each step. Fill in the one blank: inside the feedback loop,
set `feedback_msg.current_count` to the current value of `i`.


````python
import rclpy
from rclpy.node import Node
from rclpy.action import ActionServer
from example_interfaces.action import Fibonacci
import time

class CountdownActionServer(Node):
    def __init__(self):
        super().__init__('countdown_action_server')
        self._action_server = ActionServer(
            self, Fibonacci, '/countdown', self.execute_callback)

    def execute_callback(self, goal_handle):
        order = goal_handle.request.order
        feedback_msg = Fibonacci.Feedback()
        feedback_msg.partial_sequence = []

        for i in range(order, 0, -1):
            if goal_handle.is_cancel_requested:
                goal_handle.canceled()
                self.get_logger().info('Countdown cancelled!')
                return Fibonacci.Result()

            # YOUR CODE HERE — append the current count i to feedback_msg.partial_sequence
            self.get_logger().info(f'Counting down: {i}')
            goal_handle.publish_feedback(feedback_msg)
            time.sleep(1.0)

        goal_handle.succeed()
        result = Fibonacci.Result()
        result.sequence = feedback_msg.partial_sequence
        self.get_logger().info('Countdown complete!')
        return result

### BEGIN SOLUTION

### END SOLUTION

# --- Clean up previous run (safe to re-run this cell any number of times) ---
if 'countdown_node' in globals() and countdown_node is not None:
    try:
        countdown_node.destroy_node()
    except Exception:
        pass
if not rclpy.ok():
    rclpy.init()

countdown_node = CountdownActionServer()

print("✅ Action server running for 20 seconds.")
print("   Open a terminal NOW and run:")
print('   ros2 action send_goal /countdown example_interfaces/action/Fibonacci "{order: 5}" --feedback')
print("   You have 20 seconds — switch windows now.")

end_time = time.time() + 20
while time.time() < end_time and rclpy.ok():
    rclpy.spin_once(countdown_node, timeout_sec=0.1)

countdown_node.destroy_node()
print("✅ Done. Did you see feedback lines streaming before the final result?")
````
