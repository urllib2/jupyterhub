```python
%matplotlib widget
import numpy as np, matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

class Grid:
    def __init__(self,sz=80,res=0.05):
        self.sz=sz; self.res=res; self.lo=np.zeros((sz,sz))
    def w2c(self,x,y):
        return int(x/self.res)+self.sz//2, int(y/self.res)+self.sz//2
    def bres(self,x0,y0,x1,y1):
        cells=[]; dx=abs(x1-x0); dy=abs(y1-y0)
        sx=1 if x0<x1 else -1; sy=1 if y0<y1 else -1; e=dx-dy
        while True:
            cells.append((x0,y0))
            if x0==x1 and y0==y1: break
            e2=2*e
            if e2>-dy: e-=dy; x0+=sx
            if e2<dx: e+=dx; y0+=sy
        return cells
    def update(self,rx,ry,ang,rng,rmin=0.05,rmax=3.5):
        if rng<rmin or rng>rmax: return
        ex=rx+rng*np.cos(ang); ey=ry+rng*np.sin(ang)
        rcx,rcy=self.w2c(rx,ry); ecx,ecy=self.w2c(ex,ey)
        for cx,cy in self.bres(rcx,rcy,ecx,ecy)[:-1]:
            if 0<=cx<self.sz and 0<=cy<self.sz:
                self.lo[cy,cx]=np.clip(self.lo[cy,cx]-0.4,-5,5)
        if 0<=ecx<self.sz and 0<=ecy<self.sz:
            self.lo[ecy,ecx]=np.clip(self.lo[ecy,ecx]+0.85,-5,5)
    def prob(self): return 1/(1+np.exp(-self.lo))

def room_scan(rx,ry,ryaw):
    angles=np.linspace(-np.pi,np.pi,180)
    ranges=[]
    for a in angles:
        ga=a+ryaw; d=4.0
        if np.sin(ga)>0.01: d=min(d,(1.5-ry)/np.sin(ga))
        elif np.sin(ga)<-0.01: d=min(d,(-1.5-ry)/np.sin(ga))
        if np.cos(ga)>0.01: d=min(d,(2.0-rx)/np.cos(ga))
        elif np.cos(ga)<-0.01: d=min(d,(-2.0-rx)/np.cos(ga))
        ranges.append(min(max(d+np.random.normal(0,0.02),0.1),3.5))
    return angles+ryaw, np.array(ranges)

grid=Grid()
path=[(0,0,0),(0.3,0,0.2),(0.6,0.3,0.6),(0.4,0.6,1.2),(0,0.4,1.8)]
fig,ax=plt.subplots(figsize=(8,8))
im=ax.imshow(grid.prob(),cmap='RdYlGn',vmin=0,vmax=1,
             origin='lower',extent=[-2,2,-2,2])
rm,=ax.plot([],[],'b^',ms=12,label='Robot')
ax.set_title('Occupancy Grid'); ax.legend()
plt.colorbar(im,ax=ax,label='P(free)→P(occupied)')
sl=widgets.IntSlider(value=0,min=0,max=len(path)-1,
    description='Scans:',layout=widgets.Layout(width='500px'))
def upd(chg):
    n=chg['new']
    for i in range(n+1):
        rx,ry,ry2=path[i]; angs,rngs=room_scan(rx,ry,ry2)
        for a,r in zip(angs,rngs): grid.update(rx,ry,a,r)
    im.set_data(grid.prob())
    rx,ry,_=path[n]; rm.set_data([rx],[ry])
    ax.set_title(f'After {n+1} scan positions')
    fig.canvas.draw_idle()
sl.observe(upd,names='value'); display(sl); plt.show()
print('Move slider → add scans → watch map build')
print('GREEN=free | RED=occupied | YELLOW=uncertain')
```

## 6.1 — The voting booth analogy

Imagine each grid cell is a candidate in an election.

Every time a laser ray **passes through** a cell: vote **FREE**.
Every time a laser ray **ends at** a cell (hits something): vote **OCCUPIED**.

After 100 scans, a cell might have:
- 95 FREE votes, 0 OCCUPIED → very likely free (hallway)
- 0 FREE votes, 80 OCCUPIED → very likely obstacle (wall)
- 2 FREE, 1 OCCUPIED → uncertain (barely explored)

**Log-odds counts these votes mathematically.**
Each FREE vote adds $-0.40$. Each OCCUPIED vote adds $+0.85$.
No multiplication of tiny probabilities. No numerical underflow.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt, numpy as np

fig,axes=plt.subplots(1,3,figsize=(15,5))
fig.patch.set_facecolor('#1a1a2e')

# Panel 1: voting process
ax=axes[0]; ax.set_facecolor('#16213e'); ax.axis('off')
ax.set_xlim(0,10); ax.set_ylim(0,10)
ax.text(5,9,'A cell receives votes',ha='center',color='white',fontsize=12,fontweight='bold')
for i,(label,color,y) in enumerate([
    ('Vote: FREE','#43A047',7),('Vote: FREE','#43A047',6),
    ('Vote: FREE','#43A047',5),('Vote: OCCUPIED','#E53935',3.5)]):
    ax.add_patch(plt.Rectangle((1.5,y-0.4),7,0.8,fc=color,alpha=0.3,ec=color,lw=2))
    ax.text(5,y,label,ha='center',va='center',color=color,fontsize=10,fontweight='bold')
ax.text(5,2.2,'Log-odds: 3×(−0.40)+1×(+0.85)=−0.35',ha='center',color='white',fontsize=9)
ax.text(5,1.2,'→ P=0.41 → likely FREE',ha='center',color='#43A047',fontsize=10,fontweight='bold')
ax.set_title('Voting Process',color='white',fontsize=12)

# Panel 2: sigmoid
ax2=axes[1]; ax2.set_facecolor('#16213e')
lo=np.linspace(-6,6,300); p=1/(1+np.exp(-lo))
ax2.plot(lo,p,'#42a5f5',lw=3)
ax2.axhline(0.5,color='white',lw=1,linestyle='--',alpha=0.5)
ax2.axvline(0,color='white',lw=1,linestyle='--',alpha=0.5)
ax2.fill_between(lo,p,0.5,where=lo>0,alpha=0.3,color='#E53935',label='Occupied')
ax2.fill_between(lo,p,0.5,where=lo<0,alpha=0.3,color='#43A047',label='Free')
ax2.set_xlabel('Log-odds (l)',color='white'); ax2.set_ylabel('P(occupied)',color='white')
ax2.tick_params(colors='white'); ax2.set_title('Log-odds→Probability',color='white',fontsize=12)
ax2.legend(facecolor='#16213e',labelcolor='white')
for sp in ax2.spines.values(): sp.set_edgecolor('#555')

# Panel 3: accumulation
ax3=axes[2]; ax3.set_facecolor('#16213e')
lo_o=[0]; lo_f=[0]
for _ in range(19):
    lo_o.append(np.clip(lo_o[-1]+0.85,-5,5))
    lo_f.append(np.clip(lo_f[-1]-0.40,-5,5))
ax3.plot([1/(1+np.exp(-l)) for l in lo_o],'#E53935',lw=2.5,label='Occupied votes')
ax3.plot([1/(1+np.exp(-l)) for l in lo_f],'#43A047',lw=2.5,label='Free votes')
ax3.axhline(0.5,color='white',lw=1,linestyle='--',alpha=0.5)
ax3.set_xlabel('Number of votes',color='white'); ax3.set_ylabel('P(occupied)',color='white')
ax3.set_title('Probability converges',color='white',fontsize=12)
ax3.tick_params(colors='white'); ax3.legend(facecolor='#16213e',labelcolor='white')
for sp in ax3.spines.values(): sp.set_edgecolor('#555')

plt.suptitle('Log-odds: voting booth metaphor',color='white',fontsize=13,fontweight='bold')
plt.tight_layout(); plt.show()
```

The sigmoid converts log-odds back to probability:
$$p = \frac{1}{1+e^{-l}}$$
Same function used in neural networks — same mathematical reason.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt, numpy as np

def bresenham(x0,y0,x1,y1):
    cells=[]; dx=abs(x1-x0); dy=abs(y1-y0)
    sx=1 if x0<x1 else -1; sy=1 if y0<y1 else -1; e=dx-dy
    while True:
        cells.append((x0,y0))
        if x0==x1 and y0==y1: break
        e2=2*e
        if e2>-dy: e-=dy; x0+=sx
        if e2<dx: e+=dx; y0+=sy
    return cells

fig,ax=plt.subplots(figsize=(10,10))
fig.patch.set_facecolor('#1a1a2e'); ax.set_facecolor('#16213e')
gv=np.zeros((20,20,3))+0.08; robot=(2,2)
for ang in np.linspace(0,2*np.pi,12,endpoint=False):
    d=np.random.uniform(6,13)
    ex=int(robot[0]+d*np.cos(ang)); ey=int(robot[1]+d*np.sin(ang))
    ex=np.clip(ex,0,19); ey=np.clip(ey,0,19)
    cells=bresenham(robot[0],robot[1],ex,ey)
    for cx,cy in cells[:-1]:
        if 0<=cx<20 and 0<=cy<20: gv[cy,cx]=[0.1,0.5,0.1]
    cx,cy=cells[-1]
    if 0<=cx<20 and 0<=cy<20: gv[cy,cx]=[0.8,0.1,0.1]
ax.imshow(gv,origin='lower')
ax.plot(robot[0],robot[1],'b^',ms=15,label='Robot',zorder=5)
ax.set_title('Bresenham ray casting\n'  
             'green=FREE cells along ray | red=OBSTACLE endpoint',
             color='white',fontsize=12)
ax.legend(facecolor='#16213e',labelcolor='white')
ax.grid(True,color='#333',lw=0.5)
plt.tight_layout(); plt.show()
print('Key: cells[:-1] → FREE | cells[-1] → OCCUPIED')
```

## 6.2 — Math: log-odds update

$$l_{new} = \text{clamp}(l_{old} + l_{sensor},\; l_{min},\; l_{max})$$

where:
- $l_{occ} = +0.85$ for occupied observation
- $l_{free} = -0.40$ for free observation
- $l_{min} = -5,\; l_{max} = +5$ (prevents overconfidence)

Converting back: $p = \sigma(l) = \frac{1}{1+e^{-l}}$
