## Your A* as a Nav2 plugin

```python
# src/rosforge_planner.py
class ROSFORGEPlanner:
    def configure(self, node, name, tf, costmap_ros):
        self.costmap_ros = costmap_ros
        self.node = node

    def createPlan(self, start, goal):
        costmap = self.costmap_ros.getCostmap()
        grid = self.costmap_to_grid(costmap)     # Exercise 1
        start_cell = self.pose_to_cell(start, costmap)
        goal_cell  = self.pose_to_cell(goal, costmap)
        path_cells = self.astar(grid, start_cell, goal_cell)  # Module 07 code
        return self.cells_to_path(path_cells, costmap, start.header)

# Build and test:
# colcon build --packages-select rosforge_planner
# ros2 launch rosforge_bringup navigation.launch.py use_rosforge:=true
```

---
## What the planner plugin receives

```
Nav2 calls createPlan() with:
  start: PoseStamped in map frame — where robot is now
  goal:  PoseStamped in map frame — where to go

Costmap values:
  0        = free space
  1-99     = inflation zone (near obstacles)
  100      = lethal obstacle
  255      = unknown

Your A* should treat cells >=lethal_threshold as obstacles.
```
