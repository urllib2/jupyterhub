## What SLAM Toolbox does internally

```
/scan  ─────────────────────────────► SLAM Toolbox
/odom  ─────────────────────────────►     │
                                          │ 1. Build local occupancy grid (Module 06)
                                          │ 2. Match current scan to stored scans
                                          │ 3. Detect loop closure (seen this place before?)
                                          │ 4. Correct pose graph
                                          ▼
/map   ◄──────────────────────────── nav_msgs/OccupancyGrid
/tf    ◄──────────────────────────── map→odom transform
```

Step 1 is exactly your Module 06 code.
Steps 2-4 are what separates SLAM from pure dead reckoning.
