## 5.1 — The skeleton analogy

Your skeleton has bones — rigid bodies.
Your skeleton has joints — connections between bones.

Elbow: **revolute joint** — rotates within limits.
Shoulder: nearly spherical — multiple axes.
Skull: **fixed** to the atlas bone.

URDF describes a robot identically.
**Links** = bones. **Joints** = joints.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig,ax=plt.subplots(figsize=(14,5))
fig.patch.set_facecolor('#1a1a2e'); ax.set_facecolor('#16213e')
ax.set_xlim(0,14); ax.set_ylim(0,7); ax.axis('off')
joints=[
    (1.5,4,'fixed','skull/chassis','#78909C','⚙'),
    (4.0,4,'revolute','elbow/arm joint','#1E88E5','↺'),
    (6.5,4,'continuous','wheel (forever)','#43A047','⟳'),
    (9.0,4,'prismatic','drawer/actuator','#E53935','↕'),
    (11.5,4,'floating','drone body (6-DOF)','#F57C00','✦'),
]
ax.axhline(y=4,xmin=0.05,xmax=0.95,color='#37474F',lw=2,zorder=1)
for x,y,jtype,example,color,icon in joints:
    ax.add_patch(mpatches.Circle((x,y),0.65,fc=color,alpha=0.85,zorder=3))
    ax.text(x,y,icon,ha='center',va='center',fontsize=18,color='white',fontweight='bold',zorder=4)
    ax.text(x,y-1.1,jtype,ha='center',color=color,fontsize=11,fontweight='bold')
    ax.text(x,y-2.0,example,ha='center',color='#B0BEC5',fontsize=8)
ax.text(7,6.2,'URDF Joint Types',ha='center',color='white',fontsize=14,fontweight='bold')
ax.text(7,0.5,'Differential drive: wheels=continuous, chassis parts=fixed',
        ha='center',color='#90A4AE',fontsize=10)
plt.tight_layout(); plt.show()
```

Each joint type has a physical meaning.
Most robot links use `fixed` or `continuous`.

`robot_state_publisher` reads URDF + `/joint_states` → broadcasts all transforms to `/tf`.

```python
import subprocess
minimal='''<?xml version="1.0"?>
<robot name="diff_drive">
  <link name="base_link"><visual><geometry>
    <cylinder radius="0.1" length="0.05"/></geometry></visual></link>
  <link name="left_wheel"/>
  <link name="right_wheel"/>
  <link name="lidar_link"/>
  <joint name="left_wheel_joint" type="continuous">
    <parent link="base_link"/><child link="left_wheel"/>
    <origin xyz="0 0.08 -0.01" rpy="1.5708 0 0"/>
    <axis xyz="0 0 1"/></joint>
  <joint name="right_wheel_joint" type="continuous">
    <parent link="base_link"/><child link="right_wheel"/>
    <origin xyz="0 -0.08 -0.01" rpy="1.5708 0 0"/>
    <axis xyz="0 0 1"/></joint>
  <joint name="lidar_joint" type="fixed">
    <parent link="base_link"/><child link="lidar_link"/>
    <origin xyz="0.15 0 0.05"/></joint>
</robot>'''
import tempfile,os,yourdfpy
with tempfile.NamedTemporaryFile(mode='w',suffix='.urdf',delete=False) as f:
    f.write(minimal); tmp=f.name
mini=yourdfpy.URDF.load(tmp); os.unlink(tmp)
mini.show(viewer=vis['mini'])
print('Minimal differential drive:')
print(f'  Links: {list(mini.link_map.keys())}')
for name,j in mini.joint_map.items():
    print(f'  {name:<25} type={j.type}')
```

## 5.2 — Math: URDF defines the TF tree

Each URDF joint defines a static or dynamic transform:

$$T_{parent\to child} = T_{origin} \cdot R(q_i) \cdot T_{offset}$$

where $q_i$ is the joint state (angle for revolute, distance for prismatic).

`robot_state_publisher` evaluates this for all joints and publishes to `/tf`.
