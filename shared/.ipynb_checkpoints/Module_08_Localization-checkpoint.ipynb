```python
%matplotlib widget
import numpy as np, matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

np.random.seed(42); N=400; MAP=10
TRUE_PATH=[(2,2),(3,2.5),(4,3),(5,4),(5.5,5),(4.5,6),(3,6),(2,5)]
parts=np.random.uniform(0,MAP,(N,2))
weights=np.ones(N)/N

def predict(p,dx,dy,sig=0.25):
    p=p.copy(); p[:,0]+=dx+np.random.normal(0,sig,N)
    p[:,1]+=dy+np.random.normal(0,sig,N)
    return np.clip(p,0,MAP)

def update_w(p,tx,ty,sig=0.8):
    d=np.sqrt((p[:,0]-tx)**2+(p[:,1]-ty)**2)
    w=np.exp(-d**2/(2*sig**2))+1e-10; return w/w.sum()

def resample(p,w):
    idx=np.random.choice(N,size=N,p=w); return p[idx].copy()

states=[(parts.copy(),weights.copy())]
for i,pos in enumerate(TRUE_PATH):
    dx=(pos[0]-TRUE_PATH[i-1][0]) if i>0 else 0
    dy=(pos[1]-TRUE_PATH[i-1][1]) if i>0 else 0
    parts=predict(parts,dx,dy)
    weights=update_w(parts,pos[0],pos[1])
    parts=resample(parts,weights)
    weights=np.ones(N)/N
    states.append((parts.copy(),np.ones(N)/N))

fig,ax=plt.subplots(figsize=(8,8))
fig.patch.set_facecolor('#1a1a2e'); ax.set_facecolor('#16213e')
sc=ax.scatter([],[],c=[],cmap='YlOrRd',s=15,alpha=0.8,vmin=0,vmax=0.015)
td,=ax.plot([],[],'c*',ms=18,label='True pos',zorder=5)
ed,=ax.plot([],[],'g^',ms=12,label='Estimate',zorder=4)
pl,=ax.plot([],[],'c--',alpha=0.4,lw=1.5)
ax.set_xlim(0,MAP); ax.set_ylim(0,MAP)
ax.set_title('Particle Filter',color='white',fontsize=13)
ax.legend(facecolor='#1a1a2e',labelcolor='white')
ax.grid(True,alpha=0.2,color='#333')
sl=widgets.IntSlider(value=0,min=0,max=len(states)-1,
    description='Step:',layout=widgets.Layout(width='500px'))
def upd(chg):
    n=chg['new']; p,w=states[n]; sc.set_offsets(p)
    if n<len(TRUE_PATH):
        tx,ty=TRUE_PATH[n]
        d=np.sqrt((p[:,0]-tx)**2+(p[:,1]-ty)**2)
        wd=np.exp(-d**2/(2*0.8**2)); wd/=wd.sum()
        sc.set_array(wd); td.set_data([tx],[ty])
        ex=np.average(p[:,0],weights=wd)
        ey=np.average(p[:,1],weights=wd)
        ed.set_data([ex],[ey])
        err=np.sqrt((ex-tx)**2+(ey-ty)**2)
        pl.set_data([t[0] for t in TRUE_PATH[:n+1]],
                    [t[1] for t in TRUE_PATH[:n+1]])
        ax.set_title(f'Step {n} | Error: {err:.2f}m | N={N}',color='white')
    fig.canvas.draw_idle()
sl.observe(upd,names='value'); display(sl); plt.show()
print('Move slider. Watch 400 hypotheses converge.')
```

## 8.1 — The dark room analogy

You wake up in a completely dark room.
You have no idea where you are.

You take three steps forward — you moved, but from **where**?

You reach out and touch a wall on your left.
Now you know: you are within arm's length of a left wall.
But which wall?

You take two more steps. You touch a wall in front.
Corner. Now you know much more.

You turn on a flashlight for a moment.
You recognize the bookshelf. **Now you know exactly where you are.**

The particle filter is 400 copies of you waking up.
Each one is a different hypothesis.
Each sensor reading eliminates impossible ones.
The survivors cluster at the truth.

This is exactly what Nav2 AMCL does with your occupancy map.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt, numpy as np

fig,axes=plt.subplots(1,4,figsize=(16,4))
fig.patch.set_facecolor('#1a1a2e')
labels=['Wake up\n(uniform)','Touch left wall\n(narrows)',
        'Touch front wall\n(corner!)','See bookshelf\n(exact!)']
colors=['#546E7A','#1565C0','#2E7D32','#E65100']
for i,(ax,label,color) in enumerate(zip(axes,labels,colors)):
    ax.set_facecolor('#16213e'); ax.set_xlim(0,10); ax.set_ylim(0,10)
    ax.axis('off'); ax.set_title(label,color='white',fontsize=9)
    ax.add_patch(plt.Rectangle((1,1),8,8,fc='none',ec='#78909C',lw=2))
    if i==0:
        pts=np.random.uniform(1,9,(120,2))
        ax.scatter(pts[:,0],pts[:,1],c=color,s=8,alpha=0.5)
    elif i==1:
        x=np.random.uniform(1,2.5,80); y=np.random.uniform(1,9,80)
        ax.scatter(x,y,c=color,s=10,alpha=0.7)
        ax.plot([1,1],[1,9],color='#E53935',lw=3)
    elif i==2:
        x=np.random.uniform(1,2.5,40); y=np.random.uniform(7.5,9,40)
        ax.scatter(x,y,c=color,s=12,alpha=0.9)
        ax.plot([1,1],[1,9],'#E53935',lw=2)
        ax.plot([1,9],[9,9],'#E53935',lw=2)
    else:
        ax.scatter([2.5],[8.5],c=color,s=200,marker='*',zorder=5)
        ax.text(3.5,8.5,'exact\nposition',color='white',fontsize=8)
plt.suptitle('Dark room: belief narrows as evidence accumulates',
             color='white',fontsize=12,fontweight='bold')
plt.tight_layout(); plt.show()
```

Each panel = one Bayes filter step.

**Prediction:** you moved → particles scatter (uncertainty grows).
**Update:** sensor → impossible particles removed (certainty grows).
**Resample:** keep survivors, discard impossible.

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt, numpy as np
from sympy import symbols, latex
from IPython.display import display, Math

display(Math(r'\text{Prediction: } \overline{bel}(x_t) = '
             r'\int p(x_t|u_t,x_{t-1})\cdot bel(x_{t-1})\,dx_{t-1}'))
display(Math(r'\text{Update: } bel(x_t) = '
             r'\eta\cdot p(z_t|x_t)\cdot\overline{bel}(x_t)'))
print()
print('Particle filter approximation:')
print('  Prediction: move each particle by odometry + noise')
print('  Update:     weight each particle by sensor likelihood')
print('  Resample:   draw N new particles proportional to weights')

# filterpy Kalman reference
from filterpy.kalman import KalmanFilter
kf=KalmanFilter(dim_x=2,dim_z=1)
kf.F=np.array([[1.,1.],[0.,1.]])
kf.H=np.array([[1.,0.]])
kf.Q=np.eye(2)*0.01; kf.R=np.array([[2.0]])
kf.P*=10.; kf.x=np.array([[0.],[0.]])
np.random.seed(1); T=60
truth=np.linspace(0,6,T); measured=truth+np.random.normal(0,1.5,T)
ests=[]; uncs=[]
for z in measured:
    kf.predict(); kf.update([[z]])
    ests.append(kf.x[0,0]); uncs.append(np.sqrt(kf.P[0,0]))
fig,ax=plt.subplots(figsize=(14,5))
fig.patch.set_facecolor('#1a1a2e'); ax.set_facecolor('#16213e')
t=np.arange(T)
ax.plot(t,truth,'#42a5f5',lw=2,label='True position')
ax.plot(t,measured,'#ef5350',alpha=0.5,ms=5,marker='.',linestyle='none',label='Noisy GPS')
ax.plot(t,ests,'#66bb6a',lw=2,label='Kalman estimate')
u=np.array(uncs)
ax.fill_between(t,np.array(ests)-2*u,np.array(ests)+2*u,
                alpha=0.2,color='#66bb6a',label='2σ band')
ax.set_title('filterpy Kalman — 1D position tracking',color='white',fontsize=13)
ax.legend(facecolor='#1a1a2e',labelcolor='white')
ax.tick_params(colors='white')
for sp in ax.spines.values(): sp.set_edgecolor('#555')
plt.tight_layout(); plt.show()
print(f'Final error: {abs(truth[-1]-ests[-1]):.3f}m')
print()
print('Why particle filter not Kalman for robot localization?')
print('  → Motion model is non-linear (sin/cos of heading)')
print('  → Kalman requires linear Gaussian — PF requires nothing')
print('  → PF can represent multimodal belief (robot might be HERE or THERE)')
```

## 8.2 — Math: Gaussian sensor model

Each particle's weight after sensor reading $z$ at true position $(z_x, z_y)$:

$$w_i = \exp\left(-\frac{d_i^2}{2\sigma^2}\right)$$

where $d_i = \|\mathbf{x}_i - (z_x,z_y)\|_2$ is the distance from particle
$i$ to the sensor measurement.

Then normalize: $\tilde{w}_i = w_i / \sum_j w_j$
