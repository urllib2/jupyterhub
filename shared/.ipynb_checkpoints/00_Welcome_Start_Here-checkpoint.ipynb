## 🔭 What you will build

Every self-driving robot does four things in a continuous loop. In this course, you build each one:

1. **Sense** — Read LiDAR scans and wheel odometry on ROS 2 topics.
2. **Map** — Build a 2D floorplan with `slam_toolbox` while you drive.
3. **Plan** — Let Nav2 find the shortest, safest path on the costmap.
4. **Drive** — Send velocity commands to the robot and watch Bumperbot follow them.

Each lesson adds one block to this loop. By Lesson 10, the loop closes.

## 🏁 The final project — where it leads

In Lesson 10, you get a simulated world and a start pose. You will:

1. Drive Bumperbot manually to map the world with `slam_toolbox`.
2. Clean the map and save it as a Nav2-ready map.
3. Launch Nav2, click a `Nav2 Goal` in RViz, and watch Bumperbot plan and drive there — and back — without hitting a wall.

No teleop. The robot finds its own way on a map it built. That is the video you just watched.

## 🔁 How each lesson works

Long lectures kill momentum. Every lesson in this course follows a strict engineering loop:

1. **Do it** — Run the commands and see Bumperbot move.
2. **Explore** — Run `ros2 topic` and `ros2 node` and observe the data before any theory.
3. **Explain** — Read exactly why it worked.
4. **Challenge** — Break or change one parameter and explain what happened.

Short. Hands-on. No long lecture before you try.

## ✅ Get ready

For Lessons 01 through 04, we stay right here in **JupyterLab**. You will write Python nodes and test ROS 2 concepts directly in your browser.

You do not need a separate desktop for the first lessons. The JupyterLab terminal is enough.

### Step 1 — Open a terminal

In JupyterLab, go to:

**File → New → Terminal**

A new terminal tab opens. Keep it open.

### Step 2 — Check your ROS 2 environment

Run these commands one by one:

```bash
source /opt/ros/jazzy/setup.bash
```

```bash
echo $ROS_DISTRO
```

You should see:

```bash
jazzy
```

Then check the ROS 2 command line:

```bash
ros2 --help
```

You should see something like:

```bash
usage: ros2 [-h] Call 'ros2 <command> -h' for more detailed usage.
```

And finally check Python:

```bash
python3 --version
```

You should see a Python 3 version, for example:

```bash
Python 3.12.3
```

If those commands work, your environment is ready.

Terminal practice with two terminals, `talker`, and `listener` happens in **Lesson 01**, right when you first need it.
