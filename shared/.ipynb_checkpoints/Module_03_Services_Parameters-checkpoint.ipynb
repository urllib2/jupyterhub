## 3.1 — The vending machine analogy

Running water flows continuously.
You do not ask for it. It just flows.
That is a **topic**.

A vending machine is different.
You press a button (request).
You wait.
A product drops (response).
That is a **service**.

**When?** Need data continuously → topic. Need one answer → service.

**Parameters** are named values attached to a node — configuration without recompiling.
In Nav2, ALL tuning values (max speed, obstacle clearance) are parameters.

```python
import rclpy
from rclpy.node import Node
from example_interfaces.srv import AddTwoInts

class CostServer(Node):
    def __init__(self):
        super().__init__('cost_server')
        self.create_service(AddTwoInts,'/path_cost',self.compute)
    def compute(self,req,resp):
        resp.sum = req.a**2 + 10*req.b
        return resp

rclpy.init()
server=CostServer()
client_node=Node('planner')
client=client_node.create_client(AddTwoInts,'/path_cost')

req=AddTwoInts.Request(); req.a=5; req.b=3
future=client.call_async(req)
rclpy.spin_until_future_complete(client_node,future,timeout_sec=2.0)
r=future.result()
print(f"Cost(dist={req.a},obstacles={req.b}) = {req.a}²+10×{req.b} = {r.sum}")
print()

# Parameters
nav=Node('nav_config')
nav.declare_parameter('max_speed',0.5)
nav.declare_parameter('clearance',0.25)
nav.declare_parameter('robot_name','turtlebot3')
for p in ['max_speed','clearance','robot_name']:
    print(f"  {p:<20} = {nav.get_parameter(p).value}")
print()
print("These map directly to Nav2 YAML config in Module 11")

server.destroy_node(); client_node.destroy_node(); nav.destroy_node()
rclpy.shutdown()
```

## 3.2 — Math: when to use each pattern

**Asynchronous (topics):** data arrives at rate $f$ Hz.
Your callback must process each message in $< 1/f$ seconds.

**Synchronous (services):** latency $= t_{server} + t_{network}$.
Do not call services in time-critical loops — use topics instead.
