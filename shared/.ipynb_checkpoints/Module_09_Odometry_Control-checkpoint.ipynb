```python
%matplotlib widget
import numpy as np, matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

TARGET=np.pi/3

def sim_pid(Kp,Ki,Kd,dt=0.05,steps=120):
    h=0.0; integral=0.0; prev_err=0.0; hs=[h]
    for _ in range(steps):
        e=TARGET-h; integral+=e*dt
        deriv=(e-prev_err)/dt; prev_err=e
        w=np.clip(Kp*e+Ki*integral-Kd*deriv,-5,5)
        h+=w*dt; hs.append(h)
    return np.degrees(np.array(hs))

fig,ax=plt.subplots(figsize=(12,5))
fig.patch.set_facecolor('#1a1a2e'); ax.set_facecolor('#16213e')
t=np.linspace(0,6,121)
line,=ax.plot(t,sim_pid(1.5,0.1,0.5),'#42a5f5',lw=2.5)
ax.axhline(y=np.degrees(TARGET),color='#ef5350',linestyle='--',lw=2,label='Target')
ax.set_ylim(-30,120); ax.set_xlabel('Time (s)',color='white')
ax.set_ylabel('Heading (°)',color='white')
ax.tick_params(colors='white')
ax.legend(facecolor='#1a1a2e',labelcolor='white')
for sp in ax.spines.values(): sp.set_edgecolor('#555')
Kp_sl=widgets.FloatSlider(value=1.5,min=0,max=6,step=0.1,description='Kp:')
Ki_sl=widgets.FloatSlider(value=0.1,min=0,max=2,step=0.05,description='Ki:')
Kd_sl=widgets.FloatSlider(value=0.5,min=0,max=2,step=0.05,description='Kd:')
def on_change(_):
    d=sim_pid(Kp_sl.value,Ki_sl.value,Kd_sl.value)
    line.set_ydata(d)
    err=abs(np.degrees(TARGET)-d[-1])
    ax.set_title(f'Kp={Kp_sl.value:.1f} Ki={Ki_sl.value:.2f}'
                 f' Kd={Kd_sl.value:.2f} | error={err:.2f}°',color='white')
    fig.canvas.draw_idle()
for sl in [Kp_sl,Ki_sl,Kd_sl]: sl.observe(on_change,names='value')
display(widgets.VBox([Kp_sl,Ki_sl,Kd_sl])); plt.show()
print('Try: Kp=5,Ki=0,Kd=0 → oscillates')
print('     Kp=0.3,Ki=0,Kd=0 → too slow')
print('     Kp=2,Ki=0.3,Kd=0.8 → good balance')
```

## 9.1 — The eyes-closed driving analogy

You close your eyes and drive at exactly 60 km/h for 10 seconds.
You think: I travelled 167 metres.
But was it exactly 60? Did you turn 0.5° without noticing?
Did your right tyre slip slightly on a wet patch?

After 4 turns of a square course, your errors have accumulated.
You open your eyes — you are 30cm away from where you started.

This is **dead reckoning** — computing position from wheel encoders.
It always drifts.

**PID is the correction:** you open your eyes briefly (sensor reading),
measure how far you drifted (error),
and steer back proportionally (P), accounting for accumulated drift (I),
while damping oscillation (D).

```python
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt, numpy as np

fig,axes=plt.subplots(1,3,figsize=(16,5))
fig.patch.set_facecolor('#1a1a2e')

def sim_drive(noise_v=0,noise_w=0):
    x,y,th=0,0,0; traj=[(0,0)]
    cmds=[(0.3,0)]*33+[(0,np.pi/2)]*10; cmds*=4
    for v,w in cmds:
        v+=np.random.normal(0,noise_v)
        w+=np.random.normal(0,noise_w)
        if abs(w)<1e-6: x+=v*np.cos(th)*0.1; y+=v*np.sin(th)*0.1
        else:
            R=v/w; x+=R*(np.sin(th+w*0.1)-np.sin(th))
            y+=R*(np.cos(th)-np.cos(th+w*0.1)); th+=w*0.1
        traj.append((x,y))
    return traj

configs=[
    ('Perfect encoders',0,0,'#42a5f5'),
    ('Light noise\n(σ_v=0.5%)',0.0015,0.008,'#66bb6a'),
    ('Real encoders\n(σ_v=1%)',0.003,0.015,'#ef5350'),
]
for ax,(title,nv,nw,color) in zip(axes,configs):
    ax.set_facecolor('#16213e')
    traj=sim_drive(noise_v=nv,noise_w=nw)
    xs,ys=zip(*traj)
    ax.plot(xs,ys,color=color,lw=1.5)
    ax.plot(xs[0],ys[0],'g^',ms=12); ax.plot(xs[-1],ys[-1],'r*',ms=12)
    ax.set_aspect('equal'); ax.grid(True,alpha=0.2,color='#333')
    err=np.sqrt(xs[-1]**2+ys[-1]**2)
    ax.set_title(f'{title}\ndrift={err:.3f}m',color='white',fontsize=10)
    ax.tick_params(colors='white')
    for sp in ax.spines.values(): sp.set_edgecolor('#555')
plt.suptitle('Dead reckoning after 4 square turns — encoder noise accumulates',
             color='white',fontsize=12,fontweight='bold')
plt.tight_layout(); plt.show()
print('This is why Module 08 particle filter is necessary.')
print('Odometry→prediction step | Lidar→update step')
```

Small wheel noise accumulates over time.
After one square loop: 30cm error from a 1% speed variation.
The particle filter corrects this by using lidar readings.

## 9.2 — Math: kinematics and PID

**Differential drive kinematics:**
$$v = \frac{r(\omega_R+\omega_L)}{2} \qquad \omega = \frac{r(\omega_R-\omega_L)}{L}$$

**Dead reckoning (arc integration):**
$$\Delta x = \frac{v}{\omega}(\sin(\theta+\omega\Delta t)-\sin\theta)$$

**PID control law:**
$$u = K_p e + K_i\int e\,dt - K_d\dot{e}$$

The **minus** sign on $K_d$: derivative opposes rate of change (damping).
