## 2.1 — Build a publisher

Below is a publisher node with two blanks. Read the comments, fill in
`# YOUR CODE HERE`, then run the cell.

**What you're filling in:**
1. `self.publisher_` — call `self.create_publisher(...)` with the right
   message type, topic name, and queue size (same numbers you saw in
   `ros2 topic info /chatter --verbose` last lesson).
2. `msg.data` — set it to any string, but include `self.counter` so you
   can see it changing each second.


```python
import rclpy
from rclpy.node import Node
from std_msgs.msg import String
import time

class MyPublisher(Node):
    def __init__(self):
        super().__init__('my_publisher')
        # YOUR CODE HERE — create a publisher: topic '/chatter', type String, queue size 10
        self.publisher_ = None
        self.counter = 0
        self.timer = self.create_timer(1.0, self.timer_callback)

    def timer_callback(self):
        msg = String()
        # YOUR CODE HERE — set msg.data to a string that includes self.counter
        self.publisher_.publish(msg)
        self.get_logger().info(f'Publishing: "{msg.data}"')
        self.counter += 1

### BEGIN SOLUTION

### END SOLUTION

# --- Clean up previous run (safe to re-run this cell any number of times) ---
if 'my_pub_node' in globals() and my_pub_node is not None:
    try:
        my_pub_node.destroy_node()
    except Exception:
        pass
if not rclpy.ok():
    rclpy.init()

my_pub_node = MyPublisher()

print("✅ Publisher running for 12 seconds.")
print("   Open a terminal NOW and run: ros2 topic echo /chatter")
print("   You have 12 seconds — switch windows immediately.")

end_time = time.time() + 12
while time.time() < end_time and rclpy.ok():
    rclpy.spin_once(my_pub_node, timeout_sec=0.1)

my_pub_node.destroy_node()
print("✅ Done. Did you see your own messages streaming in the terminal?")
```

  still `None`, the cell will error with a clear message, not fail
  silently.
- Make sure `ros2 topic echo /chatter` is running in a **separate**
  terminal from any others you have open — not the same one running
  another command.
- If 12 seconds wasn't enough time to switch windows, just re-run the
  cell — it's safe to run as many times as you need.
